```python
import cv2
import numpy as np
import os
from skimage import data
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("/content/WhatsApp Image 2024-08-21 at 12.11.28_4a79e650.jpg")
img.shape
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10,2))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title("PIC")
plt.axis('off')
plt.subplot(1,2,2)
plt.imshow(img_rgb)
plt.title("Pic RGB")
plt.axis('off')
```

```python
# Step 5: convert the image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

print("Colour image shape   :", img.shape)
print("Grayscale image shape:", gray.shape)     # only height and width now

plt.figure(figsize=(5, 5))
plt.imshow(gray, cmap="gray")
plt.title("Grayscale image")
plt.axis("off")
plt.show()
```

```python
# Step 12: add noise to the image
noisy = gray.copy()

noise = np.random.randint(0, 100, gray.shape)
noisy[noise < 5]  = 0        # some pixels become black
noisy[noise > 95] = 255      # some pixels become white

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap="gray", vmin=0, vmax=255)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(noisy, cmap="gray", vmin=0, vmax=255)
plt.title("Noisy image")
plt.axis("off")

plt.show()
```

```python
# Step 13: average (box) filter - replaces each pixel by the average of a 5x5 area
average = cv2.blur(noisy, (5, 5))

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(noisy, cmap="gray", vmin=0, vmax=255)
plt.title("Noisy")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(average, cmap="gray", vmin=0, vmax=255)
plt.title("Average filter cv2.blur")
plt.axis("off")

plt.show()
```

```python
# Step 13: Median (box) filter - replaces each pixel by the average of a 5x5 area

median = cv2.medianBlur(noisy, 5)
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(noisy, cmap="gray", vmin=0, vmax=255)
plt.title("Noisy")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(median, cmap="gray", vmin=0, vmax=255)
plt.title("median filter")
plt.axis("off")

plt.show()

median = cv2.medianBlur(noisy, 3)
```

```python
# Step 14: Gaussian filter - a weighted average, nearer pixels get more importance
gaussian = cv2.GaussianBlur(noisy, (5, 5), 0)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(noisy, cmap="gray", vmin=0, vmax=255)
plt.title("Noisy")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gaussian, cmap="gray", vmin=0, vmax=255)
plt.title("Gaussian filter")
plt.axis("off")

plt.show()
```

```python
mean = 0
sigma = 25
noise = np.random.normal(mean, sigma, gray.shape)

noisy_img = gray.astype(np.float32) + noise

noisy_img = np.clip(noisy_img, 0, 255)
noisy_img = noisy_img.astype(np.uint8)

plt.imshow(noisy_img, cmap="gray")
plt.axis("off")
plt.show()
```
