```python
import matplotlib.pyplot as plt
import cv2
```

```python
image = cv2.imread("/content/spiderman.jpg")
```

```python
horizontal_flip = cv2.flip(image, 1)

vertical_flip = cv2.flip(image, 0)

plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.show()

plt.imshow(horizontal_flip)
plt.title("Horizontal Flip")
plt.axis("off")
plt.show()

plt.imshow(vertical_flip)
plt.title("Vertical Flip")
plt.axis()
```

```python
import numpy as np

height, width = image.shape[:2]
print("width: ", width)
print("Height: ", height)

k = 0.2
M = np.float32([
    [1, k, 0],
    [0, 1, 0]
])

new_width = int(width + k * height)
shared_image = cv2.warpAffine(image, M, (new_width, height))

plt.imshow(shared_image)
plt.title("Shared Image")
plt.axis("off")
plt.show()
```
